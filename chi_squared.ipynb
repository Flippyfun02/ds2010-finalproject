```python
import matplotlib.pyplot as plt
import scipy.stats as stats
import pandas as pd
```

```python
data = pd.read_csv("clean_data.csv")

data.head()
```

## Chi Squared test between Transaction Type and Fraudulent Label

H_0 = No significant relationship between transaction type and fraud
H_a = There is a significant relationship between transaction type and fraud

```python
table = data.pivot_table(
    values='Transaction_ID',
    index='Transaction_Type',
    columns='Is_Fraud',
    aggfunc='count',
    margins=True,
    margins_name='Total'
)

table
```

```python
contingency_table = pd.crosstab(data['Transaction_Type'], data['Is_Fraud'])
contingency_table
```

```python
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi2 Statistic: {chi2}")
print(f"P-value: {p}")
```

```python
df = 2
alpha = 0.05

critical_value = stats.chi2.ppf(1 - alpha, df)
critical_value
```
